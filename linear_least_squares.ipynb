```python
import numpy as np
import scipy as sc
import matplotlib.pyplot as plt

n = 10


A = -3

B = 4

g = lambda x: A / x + B

x = np.arange(1, 11, 1)
xx = np.arange(1, 11, 0.1)
g_ar = np.array([g(x[i]) for i in range(len(x))])

eps = 0.2 * np.min(g_ar)

e_ar = np.random.uniform(-eps, eps, size=len(x))

y_ar = g_ar + e_ar

Phi = np.vstack([[1 / x], [np.ones(len(x))]]).T

Q1, R1 = sc.linalg.qr(Phi, mode='economic')

bet = np.linalg.solve(R1, Q1.T @ y_ar)

Phit = Phi.T

plt.plot(xx, g(xx), color = 'gold')


plt.scatter(x, y_ar, color = 'm')
print(bet)
plt.plot(xx, bet[0] * 1 / xx + bet[1], color = 'black')

plt.grid()

plt.legend(["Функция с заданными параметрами", "Точки функции под воздействием шума", "Функция с восстановленными параметрами"])

plt.show()
```
